```python
import os
import random
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, Subset, ConcatDataset
from torchvision import datasets, transforms, models
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score
import cv2

# --- SEED INITIALIZATION ---
SEED = 42
def seed_everything(seed=42):
    random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

seed_everything(SEED)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# --- DATASET SELECTION & PREPARATION ---
SELECTED_CLASSES = [
    'Abyssinian', 'Bengal', 'Birman', 'Bombay', 'British_Shorthair',
    'beagle', 'boxer', 'chihuahua', 'pug', 'shiba_inu'
]

# Load both trainval and test splits to get the complete dataset pool
dataset_part1 = datasets.OxfordIIITPet(root='./data', split='trainval', download=True)
dataset_part2 = datasets.OxfordIIITPet(root='./data', split='test', download=True)
raw_dataset = ConcatDataset([dataset_part1, dataset_part2])

class_to_idx_raw = dataset_part1.class_to_idx
target_indices = []
for c in SELECTED_CLASSES:
    matched_idx = [v for k, v in class_to_idx_raw.items() if k.lower() == c.lower()]
    if matched_idx:
        target_indices.append(matched_idx[0])

idx_map = {old_id: new_id for new_id, old_id in enumerate(sorted(target_indices))}
print(f"Target Class Mapping: {idx_map}")

# Filter samples from the combined pool
filtered_indices = []
filtered_targets = []
for idx in range(len(raw_dataset)):
    _, label = raw_dataset[idx]
    if label in idx_map:
        filtered_indices.append(idx)
        filtered_targets.append(idx_map[label])

# Stratified Train (70%) / Val (15%) / Test (15%)
train_idx, val_test_idx, train_y, val_test_y = train_test_split(
    filtered_indices, filtered_targets, test_size=0.30, random_state=SEED, stratify=filtered_targets
)
val_idx, test_idx, val_y, test_y = train_test_split(
    val_test_idx, val_test_y, test_size=0.50, random_state=SEED, stratify=val_test_y
)

print(f"Dataset Splitted: Train={len(train_idx)}, Val={len(val_idx)}, Test={len(test_idx)}")
```

```python
# --- DATASET WRAPPERS & DATALOADERS ---
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]

# Base transformations (No Augmentation)
transforms_no_aug = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)
])

# Active data augmentation transformations
transforms_with_aug = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomResizedCrop(224, scale=(0.8, 1.0)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)
])

class CustomDataset(Dataset):
    def __init__(self, original_dataset, indices, labels, transform=None):
        self.original_dataset = original_dataset
        self.indices = indices
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, idx):
        img, _ = self.original_dataset[self.indices[idx]]
        label = self.labels[idx]
        if self.transform:
            img = self.transform(img)
        return img, label

# Initialize train, val, and test dataset structures
train_ds_no_aug = CustomDataset(raw_dataset, train_idx, train_y, transform=transforms_no_aug)
train_ds_with_aug = CustomDataset(raw_dataset, train_idx, train_y, transform=transforms_with_aug)
val_ds = CustomDataset(raw_dataset, val_idx, val_y, transform=transforms_no_aug)
test_ds = CustomDataset(raw_dataset, test_idx, test_y, transform=transforms_no_aug)

print("Dataset wrappers configured successfully!")
```

```python
# --- MODULAR TRAINING ENGINE ---
class EarlyStopping:
    def __init__(self, patience=4, min_delta=0):
        self.patience = patience
        self.min_delta = min_delta
        self.counter = 0
        self.best_loss = None
        self.early_stop = False

    def __call__(self, val_loss):
        if self.best_loss is None:
            self.best_loss = val_loss
        elif val_loss > self.best_loss - self.min_delta:
            self.counter += 1
            if self.counter >= self.patience:
                self.early_stop = True
        else:
            self.best_loss = val_loss
            self.counter = 0

def train_model(model, train_loader, val_loader, criterion, optimizer, num_epochs=15, patience=4, device='cuda'):
    early_stopping = EarlyStopping(patience=patience)
    history = {'train_loss': [], 'val_loss': [], 'val_acc': []}
    best_model_state = None
    best_val_loss = float('inf')
    start_time = time.time()

    for epoch in range(num_epochs):
        model.train()
        running_loss = 0.0
        for inputs, labels in train_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * inputs.size(0)

        epoch_train_loss = running_loss / len(train_loader.dataset)

        model.eval()
        running_val_loss = 0.0
        corrects = 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                running_val_loss += loss.item() * inputs.size(0)
                _, preds = torch.max(outputs, 1)
                corrects += torch.sum(preds == labels.data)

        epoch_val_loss = running_val_loss / len(val_loader.dataset)
        epoch_val_acc = corrects.double().item() / len(val_loader.dataset)

        history['train_loss'].append(epoch_train_loss)
        history['val_loss'].append(epoch_val_loss)
        history['val_acc'].append(epoch_val_acc)

        if epoch_val_loss < best_val_loss:
            best_val_loss = epoch_val_loss
            best_model_state = model.state_dict().copy()

        early_stopping(epoch_val_loss)
        if early_stopping.early_stop:
            print(f"Early stopping triggered at epoch {epoch+1}")
            break

    duration = time.time() - start_time
    if best_model_state is not None:
        model.load_state_dict(best_model_state)

    return model, history, len(history['train_loss']), duration

def evaluate_test(model, test_loader, device='cuda'):
    model.eval()
    all_preds = []
    all_labels = []
    with torch.no_grad():
        for inputs, labels in test_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    acc = accuracy_score(all_labels, all_preds)
    f1 = f1_score(all_labels, all_preds, average='macro')
    return acc, f1, np.array(all_labels), np.array(all_preds)
```

```python
# --- EXPERIMENT BUILDERS & EXECUTION LOOP ---
experiments_summary = []
models_histories = {}
best_model_instance = None
best_test_f1 = -1.0

def get_trainable_params(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

# DataLoaders setup
train_loader_no_aug = DataLoader(train_ds_no_aug, batch_size=32, shuffle=True)
train_loader_with_aug = DataLoader(train_ds_with_aug, batch_size=32, shuffle=True)
val_loader = DataLoader(val_ds, batch_size=32, shuffle=False)
test_loader = DataLoader(test_ds, batch_size=32, shuffle=False)

criterion = nn.CrossEntropyLoss()

# Define configurations
configs = [
    {"code": "B1", "strategy": "Baseline From Scratch"},
    {"code": "B2", "strategy": "Feature Extraction"},
    {"code": "B3", "strategy": "Progressive Fine-Tuning"},
    {"code": "B4_NoAug", "strategy": "B3 without Augmentation"},
    {"code": "B4_WithAug", "strategy": "B3 with Augmentation"},
    {"code": "B5", "strategy": "Alternative Architecture (EfficientNet-B0)"}
]

for cfg in configs:
    print(f"\nRunning Experiment: {cfg['code']} - {cfg['strategy']}")
    seed_everything(SEED)

    # Model Initialization
    if cfg['code'] in ["B1", "B2", "B3", "B4_NoAug", "B4_WithAug"]:
        weights = None if cfg['code'] == "B1" else models.ResNet18_Weights.IMAGENET1K_V1
        model = models.resnet18(weights=weights)
        model.fc = nn.Linear(model.fc.in_features, 10)
    else:
        model = models.efficientnet_b0(weights=models.EfficientNet_B0_Weights.IMAGENET1K_V1)
        model.classifier[1] = nn.Linear(model.classifier[1].in_features, 10)

    # Parameter Freezing and Optimizer Configuration
    if cfg['code'] == "B1":
        optimizer = optim.AdamW(model.parameters(), lr=1e-3)
        lr_str = "1e-3"
    elif cfg['code'] == "B2":
        for param in model.parameters():
            param.requires_grad = False
        for param in model.fc.parameters():
            param.requires_grad = True
        optimizer = optim.AdamW(model.fc.parameters(), lr=1e-3)
        lr_str = "1e-3"
    elif cfg['code'] in ["B3", "B4_NoAug", "B4_WithAug"]:
        for param in model.parameters():
            param.requires_grad = False
        for param in model.layer4.parameters():
            param.requires_grad = True
        for param in model.fc.parameters():
            param.requires_grad = True
        optimizer = optim.AdamW([
            {'params': model.layer4.parameters(), 'lr': 1e-4},
            {'params': model.fc.parameters(), 'lr': 1e-3}
        ])
        lr_str = "l4:1e-4, fc:1e-3"
    elif cfg['code'] == "B5":
        for param in model.parameters():
            param.requires_grad = False
        for param in model.features[7].parameters():
            param.requires_grad = True
        for param in model.classifier.parameters():
            param.requires_grad = True
        optimizer = optim.AdamW([
            {'params': model.features[7].parameters(), 'lr': 1e-4},
            {'params': model.classifier.parameters(), 'lr': 1e-3}
        ])
        lr_str = "b7:1e-4, cls:1e-3"

    # Loader routing
    loader = train_loader_with_aug if cfg['code'] == "B4_WithAug" else train_loader_no_aug

    model = model.to(device)
    trainable_params = get_trainable_params(model)

    # Train & Evaluate
    trained_model, history, epochs, duration = train_model(
        model, loader, val_loader, criterion, optimizer, num_epochs=15, patience=4, device=device
    )
    val_acc = history['val_acc'][-1]
    test_acc, test_f1, _, _ = evaluate_test(trained_model, test_loader, device=device)

    # Tracking
    models_histories[cfg['code']] = history
    experiments_summary.append({
        "Code": cfg['code'],
        "Strategy": cfg['strategy'],
        "Trainable Params": trainable_params,
        "Learning Rate": lr_str,
        "Epoch Converged": epochs,
        "Duration (s)": round(duration, 2),
        "Val Accuracy": round(val_acc, 4),
        "Test Macro F1": round(test_f1, 4)
    })

    # Retain B3 explicitly for deliverables
    if cfg['code'] == "B3":
        best_model_instance = trained_model
```

```python
# --- PART C: FINAL DELIVERABLES & EVALUATION ---
import json
import matplotlib.colors as mcolors

# 1. Print Complete Summary Table
df_summary = pd.DataFrame(experiments_summary)
print("=== EXPERIMENT PERFORMANCE SUMMARY ===")
display(df_summary)

# Export Results
df_summary.to_csv('metrics_summary.csv', index=False)
with open('experiment_results.json', 'w') as f:
    json.dump(experiments_summary, f, indent=4)
print("\nMetrics successfully exported to 'metrics_summary.csv' and 'experiment_results.json'.\n")

# 2. Side-by-Side Loss Curves for B1, B2, and B3
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for i, code in enumerate(['B1', 'B2', 'B3']):
    hist = models_histories[code]
    axes[i].plot(hist['train_loss'], label='Train Loss', marker='o')
    axes[i].plot(hist['val_loss'], label='Val Loss', marker='s')
    axes[i].set_title(f"{code} Loss Curves")
    axes[i].set_xlabel("Epoch")
    axes[i].set_ylabel("Loss")
    axes[i].legend()
    axes[i].grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# 3. Failure Analysis (C3) - 2x3 Grid of Misclassified Images using best model (B3)
best_model_instance.eval()
misclassified = []
with torch.no_grad():
    for idx in range(len(test_ds)):
        img, true_label = test_ds[idx]
        input_tensor = img.unsqueeze(0).to(device)
        outputs = best_model_instance(input_tensor)
        probs = torch.softmax(outputs, dim=1)
        conf, pred_label = torch.max(probs, dim=1)

        if pred_label.item() != true_label:
            misclassified.append({
                'index': idx,
                'true': SELECTED_CLASSES[true_label],
                'pred': SELECTED_CLASSES[pred_label.item()],
                'conf': conf.item()
            })
        if len(misclassified) >= 6:
            break

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
for i, sample in enumerate(misclassified):
    ax = axes[i // 3, i % 3]
    img, _ = test_ds[sample['index']]
    # De-normalize image for visualization
    img_np = img.permute(1, 2, 0).numpy()
    img_np = img_np * np.array(IMAGENET_STD) + np.array(IMAGENET_MEAN)
    img_np = np.clip(img_np, 0, 1)

    ax.imshow(img_np)
    ax.set_title(f"True: {sample['true']}\nPred: {sample['pred']} ({sample['conf']:.2f})")
    ax.axis('off')
plt.tight_layout()
plt.suptitle("Failure Analysis - Misclassified Samples from Model B3", y=1.02, fontsize=16)
plt.show()

# 4. Grad-CAM Visualization (C5) - ResNet-18 before vs after progressive fine-tuning
class GradCAM:
    def __init__(self, model, target_layer):
        self.model = model
        self.target_layer = target_layer
        self.gradients = None
        self.activations = None
        self.hook_layers()

    def hook_layers(self):
        def forward_hook(module, input, output):
            self.activations = output
        def backward_hook(module, grad_input, grad_output):
            self.gradients = grad_output[0]
        self.target_layer.register_forward_hook(forward_hook)
        self.target_layer.register_full_backward_hook(backward_hook)

    def generate_heatmap(self, input_tensor, class_idx):
        self.model.zero_grad()
        output = self.model(input_tensor)
        loss = output[0, class_idx]
        loss.backward()

        grads = self.gradients.cpu().data.numpy()[0]
        activations = self.activations.cpu().data.numpy()[0]

        weights = np.mean(grads, axis=(1, 2))
        cam = np.zeros(activations.shape[1:], dtype=np.float32)

        for i, w in enumerate(weights):
            cam += w * activations[i]

        cam = np.maximum(cam, 0)
        cam = cv2.resize(cam, (224, 224))
        cam = cam - np.min(cam)
        if np.max(cam) != 0:
            cam = cam / np.max(cam)
        return cam

# Generate model instances for visualization
model_untuned = models.resnet18(weights=models.ResNet18_Weights.IMAGENET1K_V1)
model_untuned.fc = nn.Linear(model_untuned.fc.in_features, 10)
model_untuned.to(device).eval()

# Take a single validation set image
sample_img, sample_lbl = val_ds[0]
input_tensor = sample_img.unsqueeze(0).to(device)

# Before tuning heatmap
gradcam_untuned = GradCAM(model_untuned, model_untuned.layer4)
heatmap_untuned = gradcam_untuned.generate_heatmap(input_tensor, sample_lbl)

# After fine-tuning heatmap (using saved best B3 model instance)
gradcam_tuned = GradCAM(best_model_instance, best_model_instance.layer4)
heatmap_tuned = gradcam_tuned.generate_heatmap(input_tensor, sample_lbl)

# Visualize comparison
img_show = sample_img.permute(1, 2, 0).numpy() * np.array(IMAGENET_STD) + np.array(IMAGENET_MEAN)
img_show = np.clip(img_show, 0, 1)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(img_show)
axes[0].set_title("Original Image")
axes[0].axis('off')

axes[1].imshow(img_show)
axes[1].imshow(heatmap_untuned, cmap='jet', alpha=0.5)
axes[1].set_title("Grad-CAM (Before Fine-Tuning)")
axes[1].axis('off')

axes[2].imshow(img_show)
axes[2].imshow(heatmap_tuned, cmap='jet', alpha=0.5)
axes[2].set_title("Grad-CAM (After Fine-Tuning)")
axes[2].axis('off')

plt.tight_layout()
plt.show()
```
